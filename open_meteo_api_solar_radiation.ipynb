# Open-Meteo API with WxData

In this example, we will demonstrate using WxData with the Open-Meteo API by doing the following:

1) Downloading the latest GFS & ECMWF Solar Radiation Forecast Data

## Imports

```python
from wxdata import open_meteo_api_solar_radiation
```

### Retrieve Solar Radiation Forecasts via Open-Meteo API

```python
gfs = open_meteo_api_solar_radiation.solar_radiation_forecast(34.54, 
                                                             -115.34, 
                                                             model='ncep gfs global 0.11/0.25')
```

```python
gfs
```

```python
ecmwf = open_meteo_api_solar_radiation.solar_radiation_forecast(34.54, 
                                                             -115.34, 
                                                             model='ecmwf ifs hres 9km')
```
