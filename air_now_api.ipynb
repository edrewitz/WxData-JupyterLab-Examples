# Air Now API with WxData

In this example, we will demonstrate using WxData with the Air Now API by doing the following:

1) Downloading the latest Air Quality Observations Within A Bounding Box:
   
   i) PM2.5
   
   ii) Ozone


## Imports

```python
from wxdata import air_now_observations
```

**IMPORTANT**

Air Now API uses an API KEY. It is good cyber security practice to save the API Key into a .txt file and read the API Key in from the file.

WxData allows users to easily do this, though they need to store 1 key per file. If the user wants to manage their API Keys their own way, they can read in the
key from the file and pass it in by setting `api_key=API_KEY` and set `read_in_key_from_path=False`. If `read_in_key_from_path=True` then `api_key={full_path_to_api_key_text_file}`. 

```python
pm25 = air_now_observations.get_current_data_bounding_box(api_key=r'C:\Users\drewi\tests\keys\air_now_api_key.txt',
                                                          western_bound=-124.205070,
                                                          eastern_bound=-114.5,
                                                          southern_bound=30.21,
                                                          northern_bound=36.54)
```

```python
pm25
```

```python
o3 = air_now_observations.get_current_data_bounding_box(api_key=r'C:\Users\drewi\tests\keys\air_now_api_key.txt',
                                                          western_bound=-124.205070,
                                                          eastern_bound=-114.5,
                                                          southern_bound=30.21,
                                                          northern_bound=36.54,
                                                          parameter='ozone')
```
